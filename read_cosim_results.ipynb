```python
import pandas as pd
import os
import glob
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import matplotlib
sns.set_theme()
#sns.set(style='whitegrid')
%config InlineBackend.figure_format = 'svg'
```

```python
# read stuff from Transmission folder
print(os.getcwd())
def tsplot(ax, data,**kw):
    x = np.arange(data.shape[1])
    est = np.mean(data, axis=0)
    est_min = np.min(data,axis=0)
    est_max = np.max(data,axis=0)
    sd = np.std(data, axis=0)
    cis = (est - sd, est + sd)
    cis = (est_min, est_max)
    ax.fill_between(x,cis[0],cis[1],alpha=0.2, **kw)
    ax.plot(x,est,**kw)
    ax.margins(x=0)
```

```python
path_to_time = os.getcwd()+ '\\Transmission\dae_time.csv'
dae_time = pd.read_csv(path_to_time)
path_to_ACE = os.getcwd()+'\Transmission\dae_ACE.csv'
dae_ACE = pd.read_csv(path_to_ACE)
path_to_frequency = os.getcwd()+'\Transmission\dae_frequency.csv'
dae_frequency = pd.read_csv(path_to_frequency)
path_to_voltage_PQ_3 = os.getcwd()+'\Transmission\\voltage_PQ_3.csv'
voltage_PQ_3 = pd.read_csv(path_to_voltage_PQ_3)
path_to_voltage_all = os.getcwd()+'\Transmission\\voltage_all.csv'
voltage_all = pd.read_csv(path_to_voltage_all)
```

```python
fig, ax5 = plt.subplots()
ax5.plot(dae_time, voltage_all)
ax5.set_ylabel('Voltage (pu)')
ax5.set_xlabel('time (s)')
ax5.set_title("Transmission all bus voltages")
```
