```python
import pandas as pd
import os
import glob
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import matplotlib
sns.set(style='whitegrid')
```

```python
fdname = '37Bus.log'
ftname = 'TransmissionSim.log'
```

```python
with open(ftname) as fp:
    lines = fp.read().splitlines()
    
```

```python
p_list = []
for line in lines:
    if 'Receive active power at time' in line:
        p = line.split(' ')[-2]
        p_list.append(p)
        
```

```python
p_list = p_list[1:]
p_df = pd.DataFrame(p_list)
```

```python
ax = p_df.astype(float).plot()
ax.set_title('Power (pu) received in Transmission')
```

```python
with open(fdname) as fp:
    lines = fp.read().splitlines()
v_list = []
for line in lines:
    if 'Receive voltage mag' in line:
        v = line.split(' ')[-2]
        v_list.append(v)   
```

```python
v_list = v_list[1:]
v_df = pd.DataFrame(v_list)
```

```python
ax2 = v_df.astype(float).plot()
ax2.set_title('Voltage received in Distribution')
```
