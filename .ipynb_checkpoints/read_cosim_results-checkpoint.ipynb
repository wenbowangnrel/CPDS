```python
import pandas as pd
import os
import glob
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import matplotlib
sns.set_theme()
#sns.set(style='whitegrid')
%config InlineBackend.figure_format = 'svg'
```

```python
# read stuff from Transmission folder
os.getcwd()
```

```python
path_to_time = os.getcwd()+ '\\Transmission\dae_time.csv'
dae_time = pd.read_csv(path_to_time)
path_to_ACE = os.getcwd()+'\Transmission\dae_ACE.csv'
dae_ACE = pd.read_csv(path_to_ACE)
path_to_frequency = os.getcwd()+'\Transmission\dae_frequency.csv'
dae_frequency = pd.read_csv(path_to_frequency)
path_to_voltage_PQ_3 = os.getcwd()+'\Transmission\\voltage_PQ_3.csv'
voltage_PQ_3 = pd.read_csv(path_to_voltage_PQ_3)
```

```python
fig, ax1 = plt.subplots()
ax1.plot(dae_time, dae_ACE)
ax1.set_ylabel('ACE (MW)')
ax1.set_xlabel('time (s)')
```

```python
fig, ax2 = plt.subplots()
ax2.plot(dae_time, dae_frequency)
ax2.set_ylabel('Frequency (Hz)')
ax2.set_xlabel('time (s)')
```

```python
fig, ax3 = plt.subplots()
ax3.plot(dae_time, voltage_PQ_3) # at the load bus PQ_3 (34bus feeder head)
ax3.set_ylabel('Voltage (pu)')
ax3.set_xlabel('time (s)')
```
