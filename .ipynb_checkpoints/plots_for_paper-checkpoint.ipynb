```python
import pandas as pd
import os
import glob
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import matplotlib
#sns.set_theme()
#sns.set(style='white')
%config InlineBackend.figure_format = 'svg'
```

```python
path_to_project = os.getcwd()

path_to_TransmissionSimResults = os.path.join(path_to_project, 'Transmission', 'simulation_results')
fname = 'freq_ACE_df_12_31_2020_15_28_40.csv'

freq_ACE_df = pd.read_csv(os.path.join(path_to_TransmissionSimResults,fname),index_col=0)
fname = 'voltage_all_df_12_31_2020_15_28_40.csv'
voltage_all_df = pd.read_csv(os.path.join(path_to_TransmissionSimResults,fname),index_col=0)
fname = 'pvd1_power_df_12_31_2020_15_28_40.csv'
pvd1_power_df = pd.read_csv(os.path.join(path_to_TransmissionSimResults,fname),index_col=0)
fname = 'PVD_vsm_max_df_12_31_2020_15_28_40.csv'
PVD_vsm_max_df = pd.read_csv(os.path.join(path_to_TransmissionSimResults,fname),index_col=0)
fname = 'mppt_df_12_31_2020_15_28_40.csv'
mppt_df = pd.read_csv(os.path.join(path_to_TransmissionSimResults,fname),index_col=0)
```

```python
import matplotlib as mpl
mpl.rcParams['axes.linewidth'] = 0.5 #set the value globally


```

```python
fig, ax1 = plt.subplots(figsize=(8,3))
ax1.plot(freq_ACE_df.time_s,freq_ACE_df.frequency_Hz, label='freq',color='C0')
ax1.legend()
ax1.set_xlabel('Time (s)')
ax1.set_ylabel('Frequency (Hz)')
ax2 = ax1.twinx()
ax2.set_ylabel('DER output with AGC', color='C2')
ax2.tick_params(axis='y', labelcolor='C2')
ax2.step(pvd1_power_df.time_s, pvd1_power_df['0']*100, label='DER', color='C2')
#ax2.legend()
#ax1.grid()
```

```python
fig, ax3 = plt.subplots(figsize=(8,3))
ax3.plot(mppt_df.time_s, mppt_df['18']*0.009*100, label='P_mppt')
ax3.step(pvd1_power_df.time_s,pvd1_power_df['18']*100, label='DER output')
ax3.plot(PVD_vsm_max_df.time_s, PVD_vsm_max_df['18']*100, label='VSM max')
ax3.legend()
ax3.set_xlabel('Time (s)')
ax3.set_ylabel('MW')
```
